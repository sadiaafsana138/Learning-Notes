*Double-click here to write your answers*

**Part A Answer:**

**Sigmoid** formula: `sigmoid(z) = 1 / (1 + e^(-z))`. It squashes any number into the range **(0, 1)**.
**ReLU** formula: `ReLU(z) = max(0, z)`. It outputs `0` for negative inputs and keeps the input as it is for positive inputs.

| | Sigmoid | ReLU |
|---|---|---|
| Output range | (0, 1) | [0, infinity) |
| Shape | smooth S-curve | straight line for z > 0, flat 0 for z < 0 |
| Computation | needs `exp()` - slower | just `max(0, z)` - very fast |
| Gradient | max value is only 0.25, goes to ~0 for big +/- z | exactly 1 for z > 0, 0 for z < 0 |

**Why ReLU is preferred in hidden layers:** (1) it is very cheap to compute, (2) its gradient is `1` for positive inputs, so the error signal
does not shrink while going backward through many layers, (3) it makes the network train faster and deeper networks become trainable.

**Problem of Sigmoid in backpropagation:** the **vanishing gradient problem**. The derivative of sigmoid is `a * (1 - a)`, which is **at most 0.25**
and becomes almost 0 when the neuron is saturated (very large or very small `z`). Backprop multiplies these small numbers layer after layer (chain rule),
so the gradient reaching the early layers becomes nearly 0, their weights barely update, and the network learns very slowly or stops learning.

**Part B Answer:**

| Type | Gradient is computed using | Weight updates per epoch |
|---|---|---|
| **Batch GD** | the **whole** dataset (all 60,000 images) | 1 |
| **Stochastic GD (SGD)** | **one** sample at a time | 60,000 |
| **Mini-batch GD** | a small **batch** (e.g. 64 samples) | 60,000 / 64 = about 938 |

- **Batch GD** - Advantage: smooth, stable loss curve and exact gradient. Disadvantage: very slow and needs lots of memory for big data, only 1 update per epoch.
- **SGD** - Advantage: very fast updates, can escape shallow local minima because of noise. Disadvantage: very noisy/zig-zag loss, cannot use GPU/vectorization well.
- **Mini-batch GD** - Advantage: good balance of speed and stability, uses vectorized matrix operations (GPU friendly). Disadvantage: needs an extra hyper-parameter (batch size) and is still a little noisy.

**Recommendation for Fashion-MNIST (60,000 images): Mini-batch Gradient Descent** (we use `batch_size = 64` with `DataLoader`).
Full-batch is too slow/memory heavy for 60,000 images, and pure SGD is too noisy and slow; mini-batch gives many updates per epoch with a stable direction.

**Part C Answer:**

1. Fashion-MNIST has **10 classes** (multi-class problem). `nn.BCELoss` is only for **binary** (2-class, one probability) problems. `nn.CrossEntropyLoss` is made for multi-class:
   it takes 10 raw scores (logits) per image and the integer class label, and measures how far the predicted class probabilities are from the true class.
2. `nn.CrossEntropyLoss` **already applies `log_softmax` internally** (Softmax + Negative Log Likelihood). If we add `nn.Softmax` in the model too, softmax would be applied **twice**,
   which squashes the values, gives wrong/weaker gradients and slows or breaks the learning. So the model must output **raw logits**.
3. `loss.backward()` runs **backpropagation (autograd)**: it walks back through the computation graph using the chain rule and calculates the gradient of the loss with respect to every
   parameter (weights and biases) and stores it in `param.grad`. Then `optimizer.step()` uses these gradients to update the weights. (That is why we call `optimizer.zero_grad()` before it - gradients accumulate.)

### 2.1 Download the Dataset (1 Marks)

We will use `torchvision.datasets.FashionMNIST` to download the dataset. This is similar to how we loaded data from `.npy` files in class, but here PyTorch handles the download for us.

## Question 4: Define the Model (15 Marks)

Create a class `FashionClassifier` inheriting from `nn.Module`.

Use `nn.Sequential` with this architecture:

| Layer | Description |
|-------|-------------|
| Flatten | Convert 28×28 → 784 |
| Linear | 784 → 256 |
| ReLU | Activation |
| Linear | 256 → 128 |
| ReLU | Activation |
| Linear | 128 → 64 |
| ReLU | Activation |
| Linear | 64 → 10 |

> **Note**: Do NOT apply Softmax at the output layer.

After defining the class:
1. Create an instance of `FashionClassifier`.
2. Print the model architecture.

```python
class FashionClassifier(nn.Module):
    def __init__(self, input_size, num_classes):
        super().__init__()

        self.network = nn.Sequential(
            nn.Flatten(),                 # 28x28 -> 784
            nn.Linear(input_size, 256),   # 1st layer = 256 ta neuron
            nn.ReLU(),
            nn.Linear(256, 128),          # 2nd layer = 128 ta neuron
            nn.ReLU(),
            nn.Linear(128, 64),           # 3rd layer = 64 ta neuron
            nn.ReLU(),
            nn.Linear(64, num_classes)    # Output layer (Softmax debo na)
        )

    def forward(self, features):
        output = self.network(features)
        return output


model = FashionClassifier(input_size=28*28, num_classes=10)
print(model)
```

## Question 5: Train the Model (25 Marks)

### 5.1 Set Up Training (5 Marks)
- Loss function: `nn.CrossEntropyLoss`
- Optimizer: `optim.SGD` with `lr=0.01`
- Number of epochs: **15**

```python
criterion = nn.CrossEntropyLoss()

optimizer = optim.SGD(model.parameters(), lr=0.01)

epochs = 15
```

### 5.2 Training Loop (15 Marks)

Write a complete training loop for **15 epochs**.

For each epoch:
- Train on `train_loader` and compute training loss
- Evaluate on `val_loader` (with `torch.no_grad()`) and compute validation loss
- Print: `Epoch {epoch+1}/{epochs} | Train Loss: {:.4f} | Val Loss: {:.4f} `

Store: `train_losses`, `val_losses`, (lists, one value per epoch).

```python
train_losses = []
val_losses = []
train_accuracies = []
val_accuracies = []

for epoch in range(epochs):

    # ---------- Training ----------
    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:
        optimizer.zero_grad()

        outputs = model(images)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

        predicted = torch.argmax(outputs, dim=1)      # sobcheye boro score er class
        correct += (predicted == labels).sum().item()
        total += labels.size(0)

    train_loss = running_loss / len(train_loader)     # gor (average) loss
    train_acc = 100 * correct / total

    # ---------- Validation ----------
    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in val_loader:
            outputs = model(images)
            loss = criterion(outputs, labels)
            val_running_loss += loss.item()

            predicted = torch.argmax(outputs, dim=1)
            val_correct += (predicted == labels).sum().item()
            val_total += labels.size(0)

    val_loss = val_running_loss / len(val_loader)
    val_acc = 100 * val_correct / val_total

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    train_accuracies.append(train_acc)
    val_accuracies.append(val_acc)

    print(f'Epoch {epoch+1}/{epochs} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} ')
```

```python
# --- TEST CELL (DO NOT MODIFY) ---
assert len(train_losses) == 15, f"Expected 15 epochs, got {len(train_losses)}"
assert len(val_losses) == 15
assert train_losses[-1] < train_losses[0], "Training loss should decrease"
print("✅ Q5.2 Passed!")
print(f"   Final Train Acc: {train_accuracies[-1]:.2f}%, Final Val Acc: {val_accuracies[-1]:.2f}%")
```

### 5.3 Test Evaluation (5 Marks)

Evaluate the model on the **test set**. Print the test_accuracy (required_variable)

Also, collect all test predictions and true labels into lists called `all_preds` and `all_labels` (we'll use them for analysis later).

Print: `Test Accuracy: {:.2f}%`

```python
with torch.no_grad():
    y_pred_prob = model(X_test)

y_pred = torch.argmax(y_pred_prob, dim=1)

all_preds = y_pred.tolist()
all_labels = y_test.tolist()

test_accuracy = (y_pred == y_test).float().mean().item() * 100

print(f'Test Accuracy: {test_accuracy:.2f}%')
```

```python
# --- TEST CELL (DO NOT MODIFY) ---
assert test_accuracy > 40.0, f"Test accuracy should be > 40%, got {test_accuracy:.2f}%"
assert len(all_preds) == len(X_test), "Predictions should cover all test samples"
print(f"✅ Q5.3 Passed! Test Accuracy: {test_accuracy:.2f}%")
```
