## 1. Printing & comments

```python
# This is a single-line comment
"""This is a
multi-line string, often used as a block comment."""

print("Hello, Python!")
print("a", "b", "c", sep="-")        # custom separator
print("no newline", end=" ... ")
print("same line")
print(f"Formatted: {2 + 3 = }")       # f-string with self-documenting expr
```

## 2. Variables & types

```python
x = 10              # int
pi = 3.14159        # float
name = "Ada"        # str
is_ok = True        # bool
nothing = None      # NoneType

for v in (x, pi, name, is_ok, nothing):
    print(f"{str(v):<10} -> {type(v).__name__}")

# dynamic typing: a name can be rebound to any type
x = "now a string"
print(x, type(x).__name__)
```

## 3. Operators

```python
a, b = 7, 2
print("add", a + b)
print("sub", a - b)
print("mul", a * b)
print("truediv", a / b)     # always float
print("floordiv", a // b)   # integer division
print("mod", a % b)
print("pow", a ** b)

# comparison & logical
print(a > b and b > 0)
print(a == 7 or b == 99)
print(not False)

# augmented assignment
c = 5
c += 3
print("c =", c)
```

## 4. Strings

```python
s = "Python Programming"

print(s.lower())
print(s.upper())
print(s.replace("Python", "Snake"))
print(s.split(" "))
print(len(s))
print(s[0], s[-1])        # indexing
print(s[0:6])             # slicing
print(s[::-1])            # reverse

# raw string: backslashes stay literal
print(r"C:\Users\name")

# join + f-strings
words = ["data", "science", "rocks"]
print(" ".join(words))
n = 42
print(f"{n:05d}")         # zero-pad
print(f"{3.14159:.2f}")   # 2 decimals
```

## 5. Lists

```python
nums = [3, 1, 4, 1, 5, 9, 2, 6]

nums.append(5)
nums.insert(0, 99)
print("after add:", nums)

nums.remove(99)           # remove first matching value
last = nums.pop()         # remove & return last
print("popped:", last)

print("sorted:", sorted(nums))
print("reversed:", nums[::-1])
print("slice [1:4]:", nums[1:4])
print("max/min/sum:", max(nums), min(nums), sum(nums))
print("count of 1:", nums.count(1))
```

## 6. Tuples & sets

```python
# tuple: ordered, immutable
point = (3, 4)
x, y = point              # unpacking
print("tuple:", point, "-> x", x, "y", y)

# set: unordered, unique elements
a = {1, 2, 3, 3, 2}
b = {3, 4, 5}
print("set a:", a)
print("union:", a | b)
print("intersection:", a & b)
print("difference:", a - b)
```

## 7. Dictionaries

```python
person = {"name": "Ada", "age": 36, "field": "math"}

print(person["name"])
print(person.get("missing", "default"))   # safe access

person["age"] = 37                          # update
person["city"] = "London"                   # add
del person["field"]                         # delete

for key, value in person.items():
    print(f"{key}: {value}")

print("keys:", list(person.keys()))
print("values:", list(person.values()))
```

## 8. Conditionals

```python
score = 78

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print("grade:", grade)

# ternary expression
status = "pass" if score >= 60 else "fail"
print("status:", status)
```

## 9. Loops

```python
# for over a range
for i in range(1, 6):
    print(i, end=" ")
print()

# for over a collection with index
for idx, ch in enumerate("abc"):
    print(idx, ch)

# while loop
n = 5
factorial = 1
while n > 1:
    factorial *= n
    n -= 1
print("5! =", factorial)

# break / continue
for i in range(10):
    if i == 3:
        continue          # skip 3
    if i == 6:
        break             # stop at 6
    print(i, end=" ")
print()
```

## 10. Comprehensions

```python
# list comprehension
squares = [n**2 for n in range(1, 8)]
print("squares:", squares)

# with a condition
evens = [n for n in range(20) if n % 2 == 0]
print("evens:", evens)

# dict comprehension
sq_map = {n: n**2 for n in range(1, 6)}
print("dict:", sq_map)

# set comprehension
firsts = {word[0] for word in ["apple", "avocado", "banana"]}
print("first letters:", firsts)
```

## 11. Functions

```python
def greet(name, greeting="Hello"):
    """Return a greeting. `greeting` has a default value."""
    return f"{greeting}, {name}!"

print(greet("Ada"))
print(greet("Linus", greeting="Hi"))

# *args and **kwargs
def summarize(*args, **kwargs):
    print("positional:", args)
    print("keyword:", kwargs)

summarize(1, 2, 3, mode="fast", debug=True)

# multiple return values (returns a tuple)
def min_max(values):
    return min(values), max(values)

lo, hi = min_max([4, 8, 1, 9, 2])
print("lo", lo, "hi", hi)
```

## 12. Lambda, map, filter

```python
double = lambda x: x * 2
print("lambda:", double(21))

nums = [1, 2, 3, 4, 5, 6]
print("map:", list(map(lambda x: x**2, nums)))
print("filter:", list(filter(lambda x: x % 2 == 0, nums)))

# sorting with a key function
people = [("Ada", 36), ("Linus", 25), ("Grace", 45)]
print("by age:", sorted(people, key=lambda p: p[1]))
```

## 13. Error handling

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        return "cannot divide by zero"
    except TypeError as e:
        return f"type error: {e}"
    else:
        return result
    finally:
        pass  # runs no matter what

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "x"))

# raising your own
def set_age(age):
    if age < 0:
        raise ValueError("age cannot be negative")
    return age

try:
    set_age(-5)
except ValueError as e:
    print("caught:", e)
```

## 14. Files
Writes a file in the current working directory, reads it back, then cleans up.

```python
# write
with open("sample.txt", "w") as f:
    f.write("line 1\n")
    f.write("line 2\n")

# read all
with open("sample.txt", "r") as f:
    print("--- full read ---")
    print(f.read())

# read line by line
with open("sample.txt", "r") as f:
    for i, line in enumerate(f, 1):
        print(f"line {i}: {line.strip()}")

import os
os.remove("sample.txt")     # cleanup
print("file exists:", os.path.exists("sample.txt"))
```

## 15. Modules & imports

```python
import math
from math import sqrt, pi
import math as m

print("math.factorial(5):", math.factorial(5))
print("sqrt(144):", sqrt(144))
print("pi:", round(pi, 4))
print("alias m.ceil(4.2):", m.ceil(4.2))

# see what a module offers
print("some math names:", [n for n in dir(math) if not n.startswith("_")][:8])
```

## 16. Dates & times

```python
from datetime import datetime, timedelta, date

now = datetime.now()
print("now:", now.strftime("%Y-%m-%d %H:%M:%S"))

today = date.today()
print("today:", today)

# arithmetic with timedelta
in_one_week = today + timedelta(days=7)
print("in a week:", in_one_week)

# parse a string into a date
parsed = datetime.strptime("2025-01-15", "%Y-%m-%d")
print("parsed:", parsed.date())
print("days since:", (today - parsed.date()).days, "days")
```

## 17. Random numbers

```python
import random

random.seed(42)                       # reproducible
print("randint 1-100:", random.randint(1, 100))
print("random float:", round(random.random(), 4))
print("choice:", random.choice(["red", "green", "blue"]))
print("sample 3:", random.sample(range(1, 50), 3))

deck = list(range(1, 11))
random.shuffle(deck)
print("shuffled:", deck)
```

## 18. `any` / `all`

```python
flags = [True, True, False]
print("any:", any(flags))     # True if at least one is truthy
print("all:", all(flags))     # True only if every item is truthy

nums = [2, 4, 6, 8]
print("all even:", all(n % 2 == 0 for n in nums))
print("any > 7:", any(n > 7 for n in nums))
```
